```python
# Generate shuffled indices of length len(dataset) for all three datasets. 

import random, os, torchvision
import numpy as np

import utils

poison_method = "BadNets-white"
transform = torchvision.transforms.Compose([
        torchvision.transforms.ToTensor(),
    ])
random.seed(3456)
path_to_folder = "./shuffle_indices"
ds_names = ["CIFAR10", "gtsrb", "imagenet"]
for train in [True, False]:
    for ds_name in ds_names:
        path_to_file = os.path.join(path_to_folder, f"{ds_name}_shuffle_indices_train_{train}.npy")
        if ds_name == "CIFAR10":
            dataset = torchvision.datasets.CIFAR10("./datasets/cifar10", train=train, transform=transform) 
        elif ds_name == "gtsrb":
            dataset = torchvision.datasets.GTSRB("./datasets", 
                                                split="train" if train == True else "test", transform=transform)
        elif ds_name == "imagenet":
            dataset =  torchvision.datasets.DatasetFolder("./datasets/tiny_imagenet_200/train" if train == True else "./datasets/tiny_imagenet_200/val/images", 
                                                            transform=transform, extensions="jpeg", 
                                                            loader=utils.loader)
        lst = list(range(len(dataset)))
        random.shuffle(lst)
        print(lst[0:10])
        np.save(path_to_file, lst)
```
